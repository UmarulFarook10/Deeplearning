 # create CNN model

```python
model=models.Sequential([
    #convolution
    layers.Conv2D(filters=32,kernel_size=(3,3),activation="relu",input_shape=(32,32,3)),

    #pooling
    layers.MaxPooling2D(),

    #Another convolution
    layers.Conv2D(filters=64,kernel_size=(3,3),activation="relu"),

    #pooling 
    layers.MaxPooling2D(),

    #convter feature map to vectore
    layers.Flatten(),

    #fully connected layer
    layers.Dense(64,activation="relu"),

    #output layer
    layers.Dense(10,activation="softmax")

])
```

32x32 image height and width

3-RGB channels

Com/2D(32) creates 32 filters/kernels

MaxPooling2D() reduces the feature-map size

Conv2D(64)→ creates o filters

Flatten() converts feature maps into a 1D vector

Dense(64) fully connected layer

Dense(10)→ gives probabilities for the 10 classes

```python
model.compile(optimizer='adam',loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
model.fit(xtrain,ytrain,epochs=3)
```

```python
test_loss,test_accuracy=model.evaluate(xtest,ytest)
```

```python
print("accuracy",test_accuracy)
```
