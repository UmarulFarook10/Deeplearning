```python
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
url="https://storage.googleapis.com/download.tensorflow.org/example_images/320px-Felis_catus-cat_on_snow.jpg"
```

```python
path=tf.keras.utils.get_file("cat.jpg",url)
```

```python
image=tf.keras.utils.load_img(path,target_size=(64,64))
```

```python
tensor=tf.keras.utils.img_to_array(image)
```

```python
print(tensor.shape)
print(tensor)
```

```python
tensor=tensor[None,...]
```

```python
print("batch shape",tensor.shape)
```

```python
conv=tf.keras.layers.Conv2D(filters=1,kernel_size=3,activation='relu')
```

```python
feature_map=conv(tensor)
print(feature_map.shape)
```

```python
pool=tf.keras.layers.MaxPooling2D(pool_size=1)
pooled=pool(feature_map)
print(pooled.shape)
```

```python
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(tensor[0].astype("uint8"))
plt.title("orginal_image")
plt.axis("off")


plt.subplot(1,3,2)
plt.imshow(feature_map[0,:,:,0],cmap="gray")
plt.title("feature_map")
plt.axis("off")


plt.subplot(1,3,3)
plt.imshow(pooled[0,:,:,0],cmap="gray")
plt.title("Ater max pooling")
plt.axis("off")

plt.show
```
