# stemming 

 Stemming is a text preprocessing technique used in Natural Language Processing (NLP) to reduce different forms of a word to a common base form, called a stem.

Stemming does not necessarily produce a real English word

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
words=["dancing","playing","writing","worked","studying"]
for word in words:
    print(word,"->",stemmer.stem(word))
```

When should you use stemming?

Use stemming when:

You want fast preprocessing

Exact grammatical correctness isn't important

You are working with large amounts of text

You want to reduce vocabulary size

You are doing search/information retrieval

You are building a basic NLP classification system


Common places where stemming is used
1. Search engines

Suppose a user searches:

"playing football"

A document may contain:

"play football"

"played football"

"players were playing football"

Stemming can reduce related words:

play

playing → play

played-play

This helps the search system find documents containing different forms of the word.

# lemmatization

Lemmatization is an NLP technique that converts a word into its base or dictionary form, called a lemma.

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
lemmatizer=WordNetLemmatizer()
words=["dancing","playing","writing","worked","studying"]
for word in words:
    print(word,"->",lemmatizer.lemmatize(word))
```

pos="n"#noun

pos="v"#verb

pos="a"#adjective

pos="r"#adverb

```python
nltk.download('wordnet')
lemmatizer=WordNetLemmatizer()
words=["dancing","playing","writing","worked","studying"]
for word in words:
    print(word,"->",lemmatizer.lemmatize(word,pos="n"))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
documents=["I love AI","I love python"]
vectorizer=CountVectorizer(token_pattern=r"(?u)\b\w+\b")
x=vectorizer.fit_transform(documents)
```

# N-grams

n-grams are a way of representing group of consecutive words from a sentence

They are important because Bag of word ignotr word order,while n-grame allow us to capture some information about the order and relationship between nearby words.

An n-gram is a sequence of N consecutive words or tockens extracted from text

N=1-UNigram

N=2-Bigram

N=3-Trigram

N=3-4-gram

```python
from nltk.util import ngrams

text = "I love machine learning"
words = text.split()

bigrams = list(ngrams(words, 2))
trigrams = list(ngrams(words, 3))

print(bigrams)
print(trigrams)
```

```python
from nltk.util import ngrams

text = "I love natural language processing"
words = text.split()

unigrams = list(ngrams(words, 1))
fourgrams = list(ngrams(words, 4))

print("Unigrams:", unigrams)
print("4-grams:", fourgrams)
```

# Term Frequency -TF

how important is a word within one document

TF tells us how important a word is inside a particular document based on its frequency

tf=no of times term t / total no of words in doc d

paython is easy to learn

words=5 

tf(paythonn,d1)=1/5=0.20

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "Machine learning is a powerful technology for data analysis",
    "Machine learning algorithms can learn patterns from data",
    "Data analysis helps businesses make better decisions",
    "Deep learning is a type of machine learning technology",
    "Python is widely used for machine learning and data science"
]

vectorizer = CountVectorizer()
tf = vectorizer.fit_transform(documents)

print("Words:")
print(vectorizer.get_feature_names_out())

print("\nTF:")
print(tf.toarray())
```

# Inverse document frequency -Idf

idf measures how rare or unique a word is across a collection of documents

idf(t)=log(n/DF(t))

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "Machine learning is a powerful technology for data analysis",
    "Machine learning algorithms can learn patterns from data",
    "Data analysis helps businesses make better decisions",
    "Deep learning is a type of machine learning technology",
    "Python is widely used for machine learning and data science"
]

vectorizer = TfidfVectorizer()
vectorizer.fit(documents)

words = vectorizer.get_feature_names_out()
idf = vectorizer.idf_

print("IDF values:\n")

for word, value in zip(words, idf):
    print(f"{word:15} {value:.4f}")
```
